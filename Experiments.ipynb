# Learning-based Models for District Heating Grids

This skript runs the different experiments used for the paper "Efficient Training Data Generation for Learning-Based State Estimation in 4th generation District Heating Grids" by A. Bott, M. Beykirch and F. Steinke (2023).


## Imports

```python
import config.experiment_config as config
from config.scenarios_config import test_cases
```


Select the experiment to run by setting the corresponding boolean to True. The experiments are:
- **'IS_sampling'**: Compares the sampling times for the classical DC+NR algorithm and the proposed IS algorithm
- **'Generate_training_samples'**: Generates training samples for the learning-based state estimation
- **'Generate_test_samples'**: Generates test samples to evaluate the model performance
- **'Train_models_WMSE'**: Trains the learning-based state mappings using the weighted MSE loss
- **'Train_models_PAL'**: Trains the learning-based state mappings models using the Physics-aware sample-free loss
- **'all'**: Runs all experiments

Select the Grid to use by specifying the grid_id. The grids are:
- **'0'**: ladder4 with supplies at positon 0 and 3
- **'1'**: ladder5 with supplies at positon 0 and 4
- **'2'**: ladder6 with supplies at positon 0 and 5
- **'3'**: ladder10 with supplies at positon 0 and 9
- **'4'**: ladder16 with supplies at positon 0, 5, 10 and 15
- **'5'**: cycle4 with supplies at positon 0
- **'6'**: cycle5 with supplies at positon 0
- **'7'**: cycle6 with supplies at positon 0
- **'8'**: cycle10 with supplies at positon 0
- **'9'**: cycle12 with supplies at positon 0 and 6

```python
grid_id = 0

run_IS_sampling = False
run_Generate_training_samples = False
run_Generate_test_samples = False
run_Train_models_WMSE = False
run_Train_models_PAL = True
run_all = False
```

## Run experiments

```python
def run_experiment(tasks):
    """
    Each experiments consists of a list of tasks. Each task is a tuple of the form (task_identifier, task_params).
    The task_identifier is a function that returns a task object. The task_params are the parameters for the task_identifier.
    The task object is a child of lib_lbm.utility.task and has to implement the following methods:
    - execute: executes the task and returns a dictionary of parameters that are passed along to the next task
    - report: reports the results of the task 
    """
    pass_along = dict()
    for task_identifier, task_params in tasks:
        task = task_identifier(**task_params, **pass_along)
        print(f'execute task: {task}')
        pass_along = {**pass_along, **task.execute()}
        task.report()
        
```

```python
# select grid: 
grid = test_cases[grid_id]
print(f'Run experiments for grid: {grid["scenario"]}')

# run experiments
if run_IS_sampling or run_all:
    run_experiment(config.get_experiment_IS_sampling(grid))
if run_Generate_training_samples or run_all:
    run_experiment(config.get_experiment_Generate_training_samples(grid))
if run_Generate_test_samples or run_all:
    run_experiment(config.get_experiment_Generate_test_samples(grid))
if run_Train_models_WMSE or run_all:
    run_experiment(config.get_experiment_Train_models_WMSE(grid))
if run_Train_models_PAL or run_all:
    run_experiment(config.get_experiment_Train_models_PAL(grid))
```
